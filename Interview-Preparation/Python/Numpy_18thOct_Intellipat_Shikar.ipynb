```python
def timecal1():
    start = time.time()
    current_time=time.localtime()
    print(current_time.tm_year)
    print(current_time)
    print("DATE")
    print(time.strftime("%m/%d/%Y, %H:%M:%S", current_time))
    end =time.time()
    return end-start
timecal1()
```

```python
def timecal2():
    start = time.time()
    current_time=time.ctime()
    print(current_time)
    end =time.time()
    return end-start
timecal2()
```

```python
def decorCalci(fun):
    start = time.time()
    def wrapper(n):
        print("Wrapper Fun")
        fun(n)
        end = time.time()
        return(end-start,'time take for the execution')
    return wrapper
```

```python
@decorCalci
def lst(n):
    for i in range (n):
        print(i,"main")    
```

```python
lst(2)
```

```python
def permute(data, i, length): 
    print(i,"iteration")
    if i==length: 
        print(''.join(data) )
    else: 
        for j in range(i,length): 
            #swap
            data[i], data[j] = data[j], data[i]
            print(data[j], data[i],'ij')
#             print("123",i,j)
            permute(data, i+1, length)
            data[i], data[j] = data[j], data[i]
#             print('234',j,i)
  

string = "ABC"
n = len(string) 
data = list(string) 
permute(data, 0, n)
```
